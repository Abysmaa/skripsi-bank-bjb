## Alur Notebook

1. Tentukan `app_id` aplikasi Bank BJB di Google Play.
2. Panggil fungsi `fetch_reviews()` dari `src.scraper`.

```python
import pandas as pd

from src.scraper import fetch_reviews, save_raw_reviews
```

```python
APP_ID = "com.example.app"
REVIEW_COUNT = 1000
LANG = "id"
COUNTRY = "id"
SORT = "newest"

# Ganti APP_ID dengan ID aplikasi mobile banking Bank BJB yang benar.
```

```python
reviews = fetch_reviews(
    app_id=APP_ID,
    count=REVIEW_COUNT,
    lang=LANG,
    country=COUNTRY,
    sort=SORT,
)
save_raw_reviews(reviews)
print(f"Saved {len(reviews)} reviews to: {DATA_PATH}")
```
